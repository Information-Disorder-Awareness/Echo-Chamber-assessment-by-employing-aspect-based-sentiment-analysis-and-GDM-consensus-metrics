```python
import pandas as pd
from bertopic import BERTopic
from bertopic.vectorizers import ClassTfidfTransformer
from modules.community_detection import community_detection
```

```python
raw_data = pd.read_csv('./data/vaccination2.csv')
```

```python
communities = community_detection("./data/", "vax", comType="weightComm")
```

```python
with open("communities.txt", "w") as f:
    f.write(str(communities))
```

```python
df = raw_data.copy()
df = df[["username", "tweet"]]
df.head()
```

```python
df_com0 = df[df['username'].isin(communities[0])]
df_com1 = df[df['username'].isin(communities[1])]
```

```python
def bert_topic(docs):
    ctfidf_model = ClassTfidfTransformer(reduce_frequent_words=True)
    topic_model = BERTopic(ctfidf_model=ctfidf_model)
    topics, probs = topic_model.fit_transform(docs)
    return topic_model
```

```python
docs = df_com0.tweet.to_list()
com0 = bert_topic(docs)
```

```python
com0.save("com0", serialization="safetensors", save_ctfidf=True)
```
