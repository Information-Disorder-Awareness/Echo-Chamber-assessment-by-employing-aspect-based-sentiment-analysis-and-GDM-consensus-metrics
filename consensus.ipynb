```python
import os
import pandas as pd
import networkx as nx
from src.gdm import compute_metrics
path = ''
name = 'rot'
```

```python
def community_detection(path, name, comType="weightComm"):
    """Extracts communities from labeled graph created by echochamber2 library.
    :param comType: label of the community in the original graph weightComm or sentComm
    :return: communities represented as lists of nodes
    """
    communities = list()
    G = nx.read_gml(os.path.join(
        path, f'Graph/Final_DiGraph_{name.capitalize()}.gml'))
    print(nx.info(G))
    community_node = list()

    for node in G.nodes():
        if G.nodes[node][comType] == 0:
            community_node.append(node)

    communities.append(community_node.copy())

    community_node.clear()

    for node in G.nodes():
        if G.nodes[node][comType] == 1:
            community_node.append(node)

    communities.append(community_node.copy())

    return communities
```

```python
df = pd.read_csv("sentiment_intensity.csv")
```

```python
communities = community_detection(path, name)
```

```python
results = compute_metrics(df, communities)
```

```python
from IPython.display import HTML
res = pd.DataFrame.from_dict(results, orient="index", columns=['aspect consensus', 'community consensus diff'])
res['community consensus'] = 1 - res['community consensus diff']
display(HTML(res.to_html()))
```
